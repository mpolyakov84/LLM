```python
from datasets import load_dataset, Dataset
import csv
from pathlib import Path
```

```python

path = '../final_dataset'

# #imagefolder
data_dir_images = Path('../final_dataset/test/images')
data_dir_labels = Path('../final_dataset/test/labels')
# dataset = load_dataset("imagefolder", data_dir=path)

#save_to_disk
```

```python
#creates metadata.csv for labels in test

with (open(data_dir_images/'metadata.csv', 'w', newline='') as output_file):
    csv_writer = csv.writer(output_file)
    csv_writer.writerow(['file_name','label','x_center', 'y_center', 'width', 'height'])
    
    for file in data_dir_labels.glob('*.txt'):
        with open(file,'r') as f:
            lines = f.readlines()
            file_name = file.stem+'.jpg'
            for line in lines:
                row = [file_name]
                row.extend(line.split(' '))
                row[-1]=row[-1].strip('\n')
                csv_writer.writerow(row)
                print(row)
        f.close()        
    output_file.close()
```
