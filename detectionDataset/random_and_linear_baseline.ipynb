# Building Random and Average BaseLine

```python
from pathlib import Path
from datasets import load_dataset
from  evaluation import  evaluate_model
import random
import pandas as pd
from sklearn.linear_model import LinearRegression
import plotly.express as px
```

```python
def random_rectangle(img='Test'):
    """
    Create a random rectangle in the image 
    :param img: 
    :return: normalized coordinates for rectangle
    """
    
    x_center = random.random()
    y_center = random.random()
    width = random.random()*(min(x_center, 1 - x_center))*2
    height = random.random()*(min(y_center, 1 - y_center))*2
    return {'x_center': x_center, 'y_center': y_center, 'width': width, 'height': height}
    
```

```python
path = '../lite_dataset'
dataset = load_dataset("imagefolder",data_dir=path, streaming=True)
```

```python
evaluate_model(random_rectangle,dataset)
```

# Linear Regression with width and height  

```python
path_train = Path('../lite_dataset/train/images')
path_test = Path('../lite_dataset/test/images')
```

```python
train_df = pd.read_csv(path_train/'metadata.csv')
test_df = pd.read_csv(path_test/'metadata.csv')
```

```python
width = train_df[['width']]
height = train_df['height']

test_width = test_df[['width']]
test_height = test_df['height']
```

```python
ln=LinearRegression()
```

```python
ln.fit(width,height)
```

```python
ln.intercept_
```

```python
predictions_height = ln.predict(test_width)
```

```python
fig = px.scatter(x=test_height,y=predictions_height)
fig.show()
```

```python
def liner_regression_height(img='Test'):
    """
    Create a random rectangle in the image with liner regression for height
    :param img: 
    :return: normalized coordinates for rectangle
    """
    
    x_center = random.random()
    y_center = random.random()
    width = random.random()*(min(x_center, 1 - x_center))*2
    height = float(ln.predict([[width]])[0])
    return {'x_center': x_center, 'y_center': y_center, 'width': width, 'height': height}
```

```python
liner_regression_height()
```

```python
dataset
```

```python
evaluate_model(liner_regression_height,dataset)
```
