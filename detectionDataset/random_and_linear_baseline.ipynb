# Building Random and Average BaseLine

```python
from pathlib import Path
from datasets import load_dataset
from  evaluation_v2 import evaluate_model
import random
import pandas as pd
from sklearn.linear_model import LinearRegression
import plotly.express as px
```

```python
def random_rectangles(img='Test'):
    """
    Create a random rectangles in the image
    :param img: 
    :return: normalized coordinates for rectangle
    """
    # result = {'categories': categories, 'bbox': bbox}
    bbox = []
    num = random.randint(1, 2)
    for i in range(num):
        x_center = random.random()
        y_center = random.random()
        width = random.random()*(min(x_center, 1 - x_center))*2
        height = random.random()*(min(y_center, 1 - y_center))*2
        bbox.append([x_center, y_center, width, height])

    return {'bbox': bbox}
    
```

```python
path = ('../lite_dataset')
dataset = load_dataset("imagefolder",data_dir=path)
```

```python
evaluate_model(random_rectangles, dataset)
```
